```python
import numpy as np

array_2x2 = np.array([
    [1,2],
    [3,4]
])

print("This is my numpy 22x2 array")
print(array_2x2)
```

```python
random_3x4 = np.random.rand(3, 4)

print("this is numpy 3x4 array")
print(random_3x4)
```

```python
ones = np.ones((3, 3))
zeros = np.zeros((3, 3))

print("this is numpy arrays ones")
print(ones)

print("this is numpy arrays zero")
print(zeros)
```

```python
random_6x3 = np.random.rand(6, 3)

print("this is numpy array 6x3")
print(random_6x3)

transposed = random_6x3.T

print("this is numpy array transposed")
print(transposed)
```

```python
random = np.random.rand(2, 3)

print("this is numpy array random")
print(random)
print("")
random2 = np.random.rand(3, 4)

print("this is numpy array random2")
print(random2)
print("")


dot_product = np.dot(random, random2)

print("this is numpy array dot_product")
print(dot_product)
```
